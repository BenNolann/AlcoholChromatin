```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define the 6 bins
bins = [-np.inf, -1, -0.5, 0, 0.5, 1, np.inf]
bin_labels = [
    'M < -1', 
    '-1 < M <= -0.5', 
    '-0.5 < M <= 0', 
    '0 < M <= 0.5', 
    '0.5 < M <= 1', 
    'M > 1'
]

# Update the bin column
ctcfrad21_df['m_bin'] = pd.cut(ctcfrad21_df['M_value_CTCF'], bins=bins, labels=bin_labels)

# 2. Define Reference (The "Zero" Anchor: M between -0.5 and 0.5)
stable_mask = (ctcfrad21_df['M_value_CTCF'] > -0.5) & (ctcfrad21_df['M_value_CTCF'] <= 0.5)
stable_seqs = ctcfrad21_df[stable_mask]['sequence_CTCF'].dropna().tolist()

def get_p_g_manual(seq_list):
    if not seq_list or len(seq_list) == 0: return None
    seq_len = len(seq_list[0])
    counts = np.zeros(seq_len)
    total = len(seq_list)
    for seq in seq_list:
        for i, base in enumerate(seq):
            if i < seq_len and base == 'G': counts[i] += 1
    return counts / total

pg_ref = get_p_g_manual(stable_seqs)

# 3. Create Heatmap Matrix
# Exclude the reference bins from the plot labels
plot_bins = ['M < -1', '-1 < M <= -0.5', '0.5 < M <= 1', 'M > 1']
heatmap_data = []
row_labels = []

for label in reversed(plot_bins):
    seqs = ctcfrad21_df[ctcfrad21_df['m_bin'] == label]['sequence_CTCF'].dropna().tolist()
    if len(seqs) > 30:
        delta_p = get_p_g_manual(seqs) - pg_ref
        heatmap_data.append(delta_p)
        row_labels.append(f"{label} (n={len(seqs)})")

hm_df = pd.DataFrame(heatmap_data, index=row_labels, columns=np.arange(1, len(pg_ref) + 1))

# 4. Plotting
plt.figure(figsize=(14, 6))
sns.heatmap(hm_df, 
            cmap='RdBu_r', 
            center=0, 
            annot=False, 
            linewidths=.5, 
            cbar_kws={'label': '$\Delta$ Probability of Guanine'})

plt.title('Guanine Susceptibility Map: $\Delta P(G)$ Relative to Stable Peaks', loc='left', fontsize=15, pad=20)
plt.xlabel('Motif Position (bp)', fontsize=12)
plt.ylabel('Ethanol Sensitivity ($log_2$FC Bin)', fontsize=12)
plt.tight_layout()
plt.savefig('6groups_guanine_heatmap.pdf')
```

### Additive analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Zinc Finger segments based on 19bp motif
zf_slices = {
    'ZF8': slice(0, 3), 'ZF7': slice(3, 6), 'ZF6': slice(6, 9),
    'ZF5': slice(9, 12), 'ZF4': slice(12, 15), 'ZF3': slice(15, 18)
}

# Extract triplets into new columns
for name, slc in zf_slices.items():
    ctcfrad21_df[name] = ctcfrad21_df['sequence_CTCF'].str[slc]

# 2. Identify all unique triplets present in the data for each ZF
# We focus on the core ZFs (8 through 3)
zf_names = list(zf_slices.keys())
all_triplets = {}
for zf in zf_names:
    all_triplets[zf] = ctcfrad21_df[zf].unique()

# 3. Calculate observed vs expected for all pairwise combinations
results = []

# Iterate through every pair of ZFs
for i in range(len(zf_names)):
    for j in range(i + 1, len(zf_names)):
        zf_a, zf_b = zf_names[i], zf_names[j]
        
        # Get all combinations of triplets for these two ZFs
        for trip_a in all_triplets[zf_a]:
            for trip_b in all_triplets[zf_b]:
                
                # Observed: Mean M_value where both conditions are met
                subset = ctcfrad21_df[(ctcfrad21_df[zf_a] == trip_a) & (ctcfrad21_df[zf_b] == trip_b)]
                if len(subset) < 30: continue
                
                obs_m = subset['M_value_CTCF'].mean()
                
                # Expected: Mean M_value of A + Mean M_value of B
                mean_a = ctcfrad21_df[ctcfrad21_df[zf_a] == trip_a]['M_value_CTCF'].mean()
                mean_b = ctcfrad21_df[ctcfrad21_df[zf_b] == trip_b]['M_value_CTCF'].mean()
                exp_m = mean_a + mean_b
                
                results.append({
                    'ZF_Pair': f"{zf_a}-{zf_b}",
                    'Triplet_Pair': f"{trip_a}-{trip_b}",
                    'Observed': obs_m,
                    'Expected': exp_m
                })

# Convert to DataFrame
comparison_df = pd.DataFrame(results)
```

```python
plt.rcParams['pdf.fonttype'] = 42
# 4. Plotting the results
plt.figure(figsize=(10, 8))
plt.hexbin(
    comparison_df['Expected'], 
    comparison_df['Observed'], 
    gridsize=50, 
    cmap='viridis', 
    mincnt=1
)
plt.colorbar(label='Count of Triplet Pairs')
plt.xlabel('Expected M-value (Sum of individual triplets)')
plt.ylabel('Observed M-value (Co-occurrence)')
plt.title('Interaction Effect: Expected vs Observed M-values')
plt.axline((0, 0), slope=1, color='red', linestyle='--', label='Identity (Additive)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.savefig('additive_triplets.pdf')
plt.show()
```

```python
# Calculate the Pearson correlation coefficient between Expected and Observed M-values
correlation = comparison_df['Expected'].corr(comparison_df['Observed'], method='pearson')

print(f"Pearson correlation coefficient: {correlation}")
```

### Group by motifs on zinc finger 5 and 6

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

# 1. Classification Logic (Ensuring ZF6 and ZF5 logic is consistent)
ctcfrad21_df['ZF6'] = ctcfrad21_df['sequence_CTCF'].str.slice(6, 9).str.upper()
ctcfrad21_df['ZF5'] = ctcfrad21_df['sequence_CTCF'].str.slice(9, 12).str.upper()

def classify_protector(row):
    gca, ggg = row['ZF6'] == 'GCA', row['ZF5'] == 'GGG'
    if gca and ggg: return "GCA+GGG"
    if ggg: return "GGG"
    if gca: return "GCA"
    return "Neither (Protector)"

def classify_fault(row):
    cta, gat = row['ZF6'] == 'CTA', row['ZF5'] == 'GAT'
    if cta and gat: return "CTA+GAT"
    if gat: return "GAT"
    if cta: return "CTA"
    return "Neither (Fault)"

ctcfrad21_df['Protector_Group'] = ctcfrad21_df.apply(classify_protector, axis=1)
ctcfrad21_df['Fault_Group'] = ctcfrad21_df.apply(classify_fault, axis=1)

# 2. Metadata & High-Contrast Colorblind Palette
conditions = [
    {'title': 'CTCF: Ethanol',  'm_col': 'M_value_CTCF',  'rd_col': 'read_density_Ctl_CTCF'},
    {'title': 'RAD21: Ethanol', 'm_col': 'M_value_RAD21', 'rd_col': 'read_density_Ctl_RAD21'}
]

hue_order = ["GCA+GGG", "GGG", "GCA", "Neither (Protector)", 
             "Neither (Fault)", "CTA", "GAT", "CTA+GAT"]

palette = {
    "GCA+GGG": "#B2182B",            # Deep Red
    "GGG": "#EF8A62",                # Burnt Orange
    "GCA": "#FDDBC7",                # Very Light Peach
    "Neither (Protector)": "#F7F7F7",# Near White/Neutral
    "Neither (Fault)": "#D1E5F0",    # Very Light Blue
    "CTA": "#67A9CF",                # Medium Blue
    "GAT": "#2166AC",                # Deep Blue
    "CTA+GAT": "#053061"             # Midnight Blue
}

bin_order = ['V. High', 'High', 'Mid', 'Low', 'V. Low']

# 3. Process Relative Data — keep per-row deltas (not collapsed to mean/SE) for boxplots
all_rel_rows = []
for cond in conditions:
    if cond['m_col'] not in ctcfrad21_df.columns: continue
    bins = pd.qcut(ctcfrad21_df[cond['rd_col']], q=5, labels=bin_order)
    
    for b_label in bin_order:
        bin_data = ctcfrad21_df[bins == b_label]
        n_prot_mean = bin_data[bin_data['Protector_Group'] == 'Neither (Protector)'][cond['m_col']].mean()
        n_fault_mean = bin_data[bin_data['Fault_Group'] == 'Neither (Fault)'][cond['m_col']].mean()
        
        for h_label in hue_order:
            baseline = n_prot_mean if h_label in ("GCA+GGG", "GGG", "GCA", "Neither (Protector)") else n_fault_mean
            
            if "Neither" in h_label:
                group_vals = bin_data[bin_data['Protector_Group' if h_label == "Neither (Protector)" else 'Fault_Group'] == h_label][cond['m_col']]
            else:
                group_vals = bin_data[(bin_data['Protector_Group'] == h_label) | (bin_data['Fault_Group'] == h_label)][cond['m_col']]
            
            # Keep every individual delta (value - baseline mean), not just the aggregate
            deltas = group_vals - baseline
            for d in deltas.dropna():
                all_rel_rows.append({
                    'Condition': cond['title'], 'Bin': b_label, 'Group': h_label, 'Delta_M': d
                })

rel_df = pd.DataFrame(all_rel_rows)

# 4. Tufte-style Faceted Boxplot (1x2 grid)
fig, axes = plt.subplots(1, 2, figsize=(20, 8), sharex=True, sharey=True)
axes = axes.flatten()

for i, cond in enumerate(conditions):
    ax = axes[i]
    subset = rel_df[rel_df['Condition'] == cond['title']]
    
    sns.boxplot(
        data=subset, x='Bin', y='Delta_M', hue='Group',
        order=bin_order, hue_order=hue_order, palette=palette,
        ax=ax, linewidth=0.8, fliersize=1.5, showfliers=False,
        boxprops={'edgecolor': 'black'}, medianprops={'color': 'black'},
        whiskerprops={'color': 'black'}, capprops={'color': 'black'}
    )
    
    # Tufte-esque styling
    ax.set_title(cond['title'], loc='left', fontsize=18, fontweight='500', pad=15)
    ax.axhline(0, color='black', lw=1.0)
    ax.set_xlabel('')
    ax.set_ylabel('')
    
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
    
    ax.spines['left'].set_position(('outward', 10))
    ax.spines['bottom'].set_position(('outward', 10))
    
    ax.tick_params(axis='both', which='major', labelsize=12, length=5)
    
    # Remove per-axis legend; a single shared legend is added below
    ax.get_legend().remove()

# Labels & Legend
fig.text(0.5, 0.02, 'Occupancy Bins (Untreated RD)', ha='center', fontsize=16, fontweight='bold')
fig.text(0.04, 0.5, 'Relative Stability ($\Delta$ M-value)', va='center', rotation='vertical', fontsize=16, fontweight='bold')

handles, labels = [], []
for h_label in hue_order:
    handles.append(plt.Rectangle((0, 0), 1, 1, facecolor=palette[h_label], edgecolor='black', linewidth=0.6))
    labels.append(h_label)
fig.legend(handles, labels, loc='lower center', ncol=4, frameon=False, 
           bbox_to_anchor=(0.5, -0.08), fontsize=13)

plt.subplots_adjust(hspace=0.35, wspace=0.15)
plt.savefig("grouped_pro_anti_ctcfrad21_boxplot.pdf", bbox_inches='tight')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define the 6 bins
bins = [-np.inf, -1, -0.5, 0, 0.5, 1, np.inf]
bin_labels = [
    'M < -1', 
    '-1 < M <= -0.5', 
    '-0.5 < M <= 0', 
    '0 < M <= 0.5', 
    '0.5 < M <= 1', 
    'M > 1'
]

# Update the bin column
ctcfrad21_df['m_bin'] = pd.cut(ctcfrad21_df['M_value_CTCF'], bins=bins, labels=bin_labels)

# 2. Define Reference (The "Zero" Anchor: M between -0.5 and 0.5)
stable_mask = (ctcfrad21_df['M_value_CTCF'] > -0.5) & (ctcfrad21_df['M_value_CTCF'] <= 0.5)
stable_seqs = ctcfrad21_df[stable_mask]['sequence_CTCF'].dropna().tolist()

def get_p_g_manual(seq_list):
    if not seq_list or len(seq_list) == 0: return None
    seq_len = len(seq_list[0])
    counts = np.zeros(seq_len)
    total = len(seq_list)
    for seq in seq_list:
        for i, base in enumerate(seq):
            if i < seq_len and base == 'G': counts[i] += 1
    return counts / total

pg_ref = get_p_g_manual(stable_seqs)

# 3. Create Heatmap Matrix
# Exclude the reference bins from the plot labels
plot_bins = ['M < -1', '-1 < M <= -0.5', '0.5 < M <= 1', 'M > 1']
heatmap_data = []
row_labels = []

for label in reversed(plot_bins):
    seqs = ctcfrad21_df[ctcfrad21_df['m_bin'] == label]['sequence_CTCF'].dropna().tolist()
    if len(seqs) > 30:
        delta_p = get_p_g_manual(seqs) - pg_ref
        heatmap_data.append(delta_p)
        row_labels.append(f"{label} (n={len(seqs)})")

hm_df = pd.DataFrame(heatmap_data, index=row_labels, columns=np.arange(1, len(pg_ref) + 1))

# 4. Plotting
plt.figure(figsize=(14, 6))
sns.heatmap(hm_df, 
            cmap='RdBu_r', 
            center=0, 
            annot=False, 
            linewidths=.5, 
            cbar_kws={'label': '$\Delta$ Probability of Guanine'})

plt.title('Guanine Susceptibility Map: $\Delta P(G)$ Relative to Stable Peaks', loc='left', fontsize=15, pad=20)
plt.xlabel('Motif Position (bp)', fontsize=12)
plt.ylabel('Ethanol Sensitivity ($log_2$FC Bin)', fontsize=12)
plt.tight_layout()
plt.savefig('6groups_guanine_heatmap.pdf')
```

### Additive analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Zinc Finger segments based on 19bp motif
zf_slices = {
    'ZF8': slice(0, 3), 'ZF7': slice(3, 6), 'ZF6': slice(6, 9),
    'ZF5': slice(9, 12), 'ZF4': slice(12, 15), 'ZF3': slice(15, 18)
}

# Extract triplets into new columns
for name, slc in zf_slices.items():
    ctcfrad21_df[name] = ctcfrad21_df['sequence_CTCF'].str[slc]

# 2. Identify all unique triplets present in the data for each ZF
# We focus on the core ZFs (8 through 3)
zf_names = list(zf_slices.keys())
all_triplets = {}
for zf in zf_names:
    all_triplets[zf] = ctcfrad21_df[zf].unique()

# 3. Calculate observed vs expected for all pairwise combinations
results = []

# Iterate through every pair of ZFs
for i in range(len(zf_names)):
    for j in range(i + 1, len(zf_names)):
        zf_a, zf_b = zf_names[i], zf_names[j]
        
        # Get all combinations of triplets for these two ZFs
        for trip_a in all_triplets[zf_a]:
            for trip_b in all_triplets[zf_b]:
                
                # Observed: Mean M_value where both conditions are met
                subset = ctcfrad21_df[(ctcfrad21_df[zf_a] == trip_a) & (ctcfrad21_df[zf_b] == trip_b)]
                if len(subset) < 30: continue
                
                obs_m = subset['M_value_CTCF'].mean()
                
                # Expected: Mean M_value of A + Mean M_value of B
                mean_a = ctcfrad21_df[ctcfrad21_df[zf_a] == trip_a]['M_value_CTCF'].mean()
                mean_b = ctcfrad21_df[ctcfrad21_df[zf_b] == trip_b]['M_value_CTCF'].mean()
                exp_m = mean_a + mean_b
                
                results.append({
                    'ZF_Pair': f"{zf_a}-{zf_b}",
                    'Triplet_Pair': f"{trip_a}-{trip_b}",
                    'Observed': obs_m,
                    'Expected': exp_m
                })

# Convert to DataFrame
comparison_df = pd.DataFrame(results)
```

```python
plt.rcParams['pdf.fonttype'] = 42
# 4. Plotting the results
plt.figure(figsize=(10, 8))
plt.hexbin(
    comparison_df['Expected'], 
    comparison_df['Observed'], 
    gridsize=50, 
    cmap='viridis', 
    mincnt=1
)
plt.colorbar(label='Count of Triplet Pairs')
plt.xlabel('Expected M-value (Sum of individual triplets)')
plt.ylabel('Observed M-value (Co-occurrence)')
plt.title('Interaction Effect: Expected vs Observed M-values')
plt.axline((0, 0), slope=1, color='red', linestyle='--', label='Identity (Additive)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.savefig('additive_triplets.pdf')
plt.show()
```

```python
# Calculate the Pearson correlation coefficient between Expected and Observed M-values
correlation = comparison_df['Expected'].corr(comparison_df['Observed'], method='pearson')

print(f"Pearson correlation coefficient: {correlation}")
```

### Group by motifs on zinc finger 5 and 6

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

# 1. Classification Logic (Ensuring ZF6 and ZF5 logic is consistent)
ctcfrad21_df['ZF6'] = ctcfrad21_df['sequence_CTCF'].str.slice(6, 9).str.upper()
ctcfrad21_df['ZF5'] = ctcfrad21_df['sequence_CTCF'].str.slice(9, 12).str.upper()

def classify_protector(row):
    gca, ggg = row['ZF6'] == 'GCA', row['ZF5'] == 'GGG'
    if gca and ggg: return "GCA+GGG"
    if ggg: return "GGG"
    if gca: return "GCA"
    return "Neither (Protector)"

def classify_fault(row):
    cta, gat = row['ZF6'] == 'CTA', row['ZF5'] == 'GAT'
    if cta and gat: return "CTA+GAT"
    if gat: return "GAT"
    if cta: return "CTA"
    return "Neither (Fault)"

ctcfrad21_df['Protector_Group'] = ctcfrad21_df.apply(classify_protector, axis=1)
ctcfrad21_df['Fault_Group'] = ctcfrad21_df.apply(classify_fault, axis=1)

# 2. Metadata & High-Contrast Colorblind Palette
conditions = [
    {'title': 'CTCF: Ethanol',  'm_col': 'M_value_CTCF',  'rd_col': 'read_density_Ctl_CTCF'},
    {'title': 'RAD21: Ethanol', 'm_col': 'M_value_RAD21', 'rd_col': 'read_density_Ctl_RAD21'}
]

hue_order = ["GCA+GGG", "GGG", "GCA", "Neither (Protector)", 
             "Neither (Fault)", "CTA", "GAT", "CTA+GAT"]

palette = {
    "GCA+GGG": "#B2182B",            # Deep Red
    "GGG": "#EF8A62",                # Burnt Orange
    "GCA": "#FDDBC7",                # Very Light Peach
    "Neither (Protector)": "#F7F7F7",# Near White/Neutral
    "Neither (Fault)": "#D1E5F0",    # Very Light Blue
    "CTA": "#67A9CF",                # Medium Blue
    "GAT": "#2166AC",                # Deep Blue
    "CTA+GAT": "#053061"             # Midnight Blue
}

bin_order = ['V. High', 'High', 'Mid', 'Low', 'V. Low']

# 3. Process Relative Data — keep per-row deltas (not collapsed to mean/SE) for boxplots
all_rel_rows = []
for cond in conditions:
    if cond['m_col'] not in ctcfrad21_df.columns: continue
    bins = pd.qcut(ctcfrad21_df[cond['rd_col']], q=5, labels=bin_order)
    
    for b_label in bin_order:
        bin_data = ctcfrad21_df[bins == b_label]
        n_prot_mean = bin_data[bin_data['Protector_Group'] == 'Neither (Protector)'][cond['m_col']].mean()
        n_fault_mean = bin_data[bin_data['Fault_Group'] == 'Neither (Fault)'][cond['m_col']].mean()
        
        for h_label in hue_order:
            baseline = n_prot_mean if h_label in ("GCA+GGG", "GGG", "GCA", "Neither (Protector)") else n_fault_mean
            
            if "Neither" in h_label:
                group_vals = bin_data[bin_data['Protector_Group' if h_label == "Neither (Protector)" else 'Fault_Group'] == h_label][cond['m_col']]
            else:
                group_vals = bin_data[(bin_data['Protector_Group'] == h_label) | (bin_data['Fault_Group'] == h_label)][cond['m_col']]
            
            # Keep every individual delta (value - baseline mean), not just the aggregate
            deltas = group_vals - baseline
            for d in deltas.dropna():
                all_rel_rows.append({
                    'Condition': cond['title'], 'Bin': b_label, 'Group': h_label, 'Delta_M': d
                })

rel_df = pd.DataFrame(all_rel_rows)

# 4. Tufte-style Faceted Boxplot (1x2 grid)
fig, axes = plt.subplots(1, 2, figsize=(20, 8), sharex=True, sharey=True)
axes = axes.flatten()

for i, cond in enumerate(conditions):
    ax = axes[i]
    subset = rel_df[rel_df['Condition'] == cond['title']]
    
    sns.boxplot(
        data=subset, x='Bin', y='Delta_M', hue='Group',
        order=bin_order, hue_order=hue_order, palette=palette,
        ax=ax, linewidth=0.8, fliersize=1.5, showfliers=False,
        boxprops={'edgecolor': 'black'}, medianprops={'color': 'black'},
        whiskerprops={'color': 'black'}, capprops={'color': 'black'}
    )
    
    # Tufte-esque styling
    ax.set_title(cond['title'], loc='left', fontsize=18, fontweight='500', pad=15)
    ax.axhline(0, color='black', lw=1.0)
    ax.set_xlabel('')
    ax.set_ylabel('')
    
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
    
    ax.spines['left'].set_position(('outward', 10))
    ax.spines['bottom'].set_position(('outward', 10))
    
    ax.tick_params(axis='both', which='major', labelsize=12, length=5)
    
    # Remove per-axis legend; a single shared legend is added below
    ax.get_legend().remove()

# Labels & Legend
fig.text(0.5, 0.02, 'Occupancy Bins (Untreated RD)', ha='center', fontsize=16, fontweight='bold')
fig.text(0.04, 0.5, 'Relative Stability ($\Delta$ M-value)', va='center', rotation='vertical', fontsize=16, fontweight='bold')

handles, labels = [], []
for h_label in hue_order:
    handles.append(plt.Rectangle((0, 0), 1, 1, facecolor=palette[h_label], edgecolor='black', linewidth=0.6))
    labels.append(h_label)
fig.legend(handles, labels, loc='lower center', ncol=4, frameon=False, 
           bbox_to_anchor=(0.5, -0.08), fontsize=13)

plt.subplots_adjust(hspace=0.35, wspace=0.15)
plt.savefig("grouped_pro_anti_ctcfrad21_boxplot.pdf", bbox_inches='tight')
```
